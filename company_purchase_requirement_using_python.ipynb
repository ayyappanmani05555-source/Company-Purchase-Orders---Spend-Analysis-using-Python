```python

import pandas as pd

# Load the Excel file into a pandas DataFrame
df = pd.read_excel('purchase requirment')

# Display the first 5 rows of the DataFrame
display(df.head())
```

### Dataset Understanding

```python
# Display basic information about the DataFrame
if not df.empty:
    print(df.info())
    print("\n")
    # Display descriptive statistics
    display(df.describe(include='all'))
```

### Missing Value Handling

```python
# Check for missing values
if not df.empty:
    missing_values = df.isnull().sum()
    missing_percentage = (df.isnull().sum() / len(df)) * 100
    missing_info = pd.DataFrame({'Missing Count': missing_values, 'Missing Percentage': missing_percentage})
    display(missing_info[missing_info['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False))
```

### Handling Missing Values

```python
# Fill missing 'Supplier' with 'Unknown'
if 'Supplier' in df.columns:
    df['Supplier'].fillna('Unknown', inplace=True)
    print("Missing 'Supplier' values filled with 'Unknown'.")

# Fill missing 'Unit_Price_INR' with its median
if 'Unit_Price_INR' in df.columns:
    median_unit_price = df['Unit_Price_INR'].median()
    df['Unit_Price_INR'].fillna(median_unit_price, inplace=True)
    print(f"Missing 'Unit_Price_INR' values filled with median: {median_unit_price}.")

# Verify no more missing values in these columns
print("\nMissing values after handling:")
display(df[['Supplier', 'Unit_Price_INR']].isnull().sum())
```

### Data Type Conversion and Feature Engineering (if needed)

```python
# Check 'Month' column. Since 'Purchase_Date' is already datetime, we can ensure consistency or extract features.
# The 'Month' column is currently an object type.
# Let's verify if it's consistent with 'Purchase_Date'.

if 'Purchase_Date' in df.columns and 'Month' in df.columns:
    # Extract month name from 'Purchase_Date' for comparison
    df['Month_from_Date'] = df['Purchase_Date'].dt.strftime('%Y-%m') # Format as YYYY-MM

    # Check for consistency (after cleaning any potential whitespace in 'Month')
    df['Month'] = df['Month'].str.strip() # Remove any leading/trailing whitespace

    # Format 'Month' column to match 'Month_from_Date' for comparison
    # Assuming 'Month' is 'YYYY-MM' or similar. If not, this comparison might fail.
    # Let's print unique values from original 'Month' first to understand format.
    print("Unique values in original 'Month' column:")
    display(df['Month'].unique())

    # If 'Month' column is just month name, then it needs to be transformed before comparison.
    # For now, let's just make sure 'Purchase_Date' is used for reliable month info.
    df.drop('Month', axis=1, inplace=True)
    df.rename(columns={'Month_from_Date': 'Month'}, inplace=True)
    print("\n'Month' column updated from 'Purchase_Date'.")
    display(df[['Purchase_Date', 'Month']].head())
else:
    print("Columns 'Purchase_Date' or 'Month' not found. Skipping consistency check and update.")


print("\nDataFrame info after cleaning:")
print(df.info())
```

### Feature Engineering

```python
# Extract Year, Quarter, and Day of Week from 'Purchase_Date'
if 'Purchase_Date' in df.columns:
    df['Year'] = df['Purchase_Date'].dt.year
    df['Quarter'] = df['Purchase_Date'].dt.quarter
    df['Day_of_Week'] = df['Purchase_Date'].dt.day_name()
    print("Extracted 'Year', 'Quarter', and 'Day_of_Week' features.")
    display(df[['Purchase_Date', 'Year', 'Quarter', 'Day_of_Week']].head())
else:
    print("'Purchase_Date' column not found. Skipping feature engineering.")
```

### Data Type Optimization

```python
# Convert suitable object columns to 'category' dtype
# Identify columns that are 'object' type and have a relatively low number of unique values
categorical_cols = [col for col in df.select_dtypes(include='object').columns if df[col].nunique() < len(df) / 2]

for col in categorical_cols:
    df[col] = df[col].astype('category')
print(f"Converted {len(categorical_cols)} object columns to 'category' type.")

# Display DataFrame info after all preparations
print("\nDataFrame info after full dataset preparation:")
print(df.info())
```

### Total Purchase Amount by Department

## Exploratory Data Analysis (EDA)

### Supplier Analysis

```python
# Top suppliers by total purchase value
top_suppliers_value = df.groupby('Supplier')['Final_Purchase_Amount_INR'].sum().sort_values(ascending=False)
print("Top Suppliers by Total Purchase Value (INR):")
display(top_suppliers_value.head(10))

# Top suppliers by total quantity purchased
top_suppliers_quantity = df.groupby('Supplier')['Quantity'].sum().sort_values(ascending=False)
print("\nTop Suppliers by Total Quantity Purchased:")
display(top_suppliers_quantity.head(10))

# Most frequently used suppliers (by number of purchase orders)
most_frequent_suppliers = df.groupby('Supplier')['Purchase_ID'].count().sort_values(ascending=False)
print("\nMost Frequently Used Suppliers (by number of orders):")
display(most_frequent_suppliers.head(10))

# Average purchase value per supplier
avg_purchase_per_supplier = df.groupby('Supplier')['Final_Purchase_Amount_INR'].mean().sort_values(ascending=False)
print("\nAverage Purchase Value per Supplier (INR):")
display(avg_purchase_per_supplier.head(10))
```

```python
# Visualize top suppliers by purchase value
plt.figure(figsize=(12, 7))
sns.barplot(x=top_suppliers_value.head(10).index, y=top_suppliers_value.head(10).values, palette='Blues_d')
plt.title('Top 10 Suppliers by Total Purchase Value')
plt.xlabel('Supplier')
plt.ylabel('Total Purchase Value (INR)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Product Analysis

```python
# Most purchased products (by quantity)
most_purchased_products_quantity = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)
print("Most Purchased Products (by Quantity):")
display(most_purchased_products_quantity.head(10))

# Highest-value products (by total purchase amount)
highest_value_products = df.groupby('Product')['Final_Purchase_Amount_INR'].sum().sort_values(ascending=False)
print("\nHighest-Value Products (by Total Purchase Amount INR):")
display(highest_value_products.head(10))

# Product categories generating the highest spending (total)
highest_spending_categories = df.groupby('Category')['Final_Purchase_Amount_INR'].sum().sort_values(ascending=False)
print("\nProduct Categories Generating the Highest Spending (Total INR):")
display(highest_spending_categories.head(10))
```

```python
# Visualize highest-value products
plt.figure(figsize=(12, 7))
sns.barplot(x=highest_value_products.head(10).index, y=highest_value_products.head(10).values, palette='Reds_d')
plt.title('Top 10 Products by Total Purchase Value')
plt.xlabel('Product')
plt.ylabel('Total Purchase Value (INR)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Visualize highest spending categories
plt.figure(figsize=(10, 6))
sns.barplot(x=highest_spending_categories.index, y=highest_spending_categories.values, palette='Greens_d')
plt.title('Product Categories by Total Spending')
plt.xlabel('Category')
plt.ylabel('Total Spending (INR)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Department Analysis

```python
# Department with highest purchasing expenditure (re-display for context)
department_expenditure = df.groupby('Department')['Final_Purchase_Amount_INR'].sum().sort_values(ascending=False)
print("Department with Highest Purchasing Expenditure (INR):")
display(department_expenditure)

# Department with highest number of purchase orders
department_purchase_orders = df.groupby('Department')['Purchase_ID'].count().sort_values(ascending=False)
print("\nDepartment with Highest Number of Purchase Orders:")
display(department_purchase_orders)

# Department spending trends (Monthly)
# Ensure 'Month' is sorted correctly for time series plotting
df_sorted = df.sort_values(by='Purchase_Date')
department_monthly_spending = df_sorted.groupby(['Month', 'Department'])['Final_Purchase_Amount_INR'].sum().unstack()

plt.figure(figsize=(15, 8))
department_monthly_spending.plot(kind='line', marker='o', figsize=(15, 8))
plt.title('Monthly Spending Trends by Department')
plt.xlabel('Month')
plt.ylabel('Total Purchase Amount (INR)')
plt.xticks(rotation=45, ha='right')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Department')
plt.tight_layout()
plt.show()
```

### Regional Analysis

```python
# Highest-spending region (by City)
city_spending = df.groupby('City')['Final_Purchase_Amount_INR'].sum().sort_values(ascending=False)
print("Highest Spending Regions (Cities):")
display(city_spending.head())

# Lowest-spending region (by City)
print("\nLowest Spending Regions (Cities):")
display(city_spending.tail())

# Regional purchasing patterns (Visualize spending by city)
plt.figure(figsize=(12, 7))
sns.barplot(x=city_spending.index, y=city_spending.values, palette='viridis')
plt.title('Total Purchase Amount by City (Region)')
plt.xlabel('City')
plt.ylabel('Total Purchase Amount (INR)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## GroupBy Operations for Business-Level Summaries

### GroupBy: Supplier

```python
# GroupBy Supplier: Total Purchase Amount, Total Quantity, and Number of Orders
supplier_summary = df.groupby('Supplier').agg(
    Total_Purchase_Amount=('Final_Purchase_Amount_INR', 'sum'),
    Total_Quantity_Purchased=('Quantity', 'sum'),
    Number_of_Orders=('Purchase_ID', 'count'),
    Average_Order_Value=('Final_Purchase_Amount_INR', 'mean')
).sort_values(by='Total_Purchase_Amount', ascending=False)
print("Supplier Summary:")
display(supplier_summary)
```
